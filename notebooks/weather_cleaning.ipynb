```python
import pandas as pd 
from pathlib import Path 
```

```python
data_path = Path("../data/weather_Chicago.csv")

df = pd.read_csv(data_path)

print(df.shape)
display(df.head())
```

```python
midway_df = df[df['STATION'] == 'USW00014819'].copy()

print(midway_df.shape)
display(midway_df.head())
```

```python
print(midway_df.isnull().sum())

midway_df["TAVG"]=(midway_df["TMAX"] + midway_df["TMIN"])/2

display(midway_df.head())
```

```python
midway_df["DATE"]=pd.to_datetime(midway_df["DATE"])

midway_df["MONTH"]=midway_df["DATE"].dt.month
midway_df["DAY"]=midway_df["DATE"].dt.day

display(midway_df.head())
```

#at this point, I realized I forgot "precipitation" column, I went back on original data source and got precipitation dataset

```python
prcp_df=pd.read_csv("../data/new_precipitation.csv")

prcp_df=prcp_df[["DATE","PRCP"]]

prcp_df["DATE"]=pd.to_datetime(prcp_df["DATE"])

midway_df=pd.merge(midway_df,prcp_df,on="DATE",how="left")

display(midway_df.head())
```

```python
#midway_df=midway_df.drop(columns=['PRCP_x'])

#midway_df=midway_df.rename(columns={'PRCP_y':'PRCP'})
```

```python
midway_df=midway_df.sort_values(by=["MONTH","DAY"])

midway_df=midway_df[["MONTH","DAY","TMAX","TMIN","TAVG","PRCP"]]

midway_df=midway_df.reset_index(drop=True)

display(midway_df.head())
```
